```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 1. Pairplot
df = pd.read_csv('./datasets/plant_health.csv')
df.drop(columns=['plant_id', 'nutrient_level', 'soil_ph'],
        inplace=True, errors='ignore')  # Drop unnecessary columns
sns.pairplot(df, hue='health', diag_kind='kde', palette='Set1',
             height=2.5, markers='o', plot_kws={'alpha': 0.5})
plt.suptitle('Pairplot of Plant Health Dataset', y=1.02)
plt.subplots_adjust(top=0.9, right=0.9)
plt.tight_layout()
plt.show()
```

```python
# 2. Heatmap
df = pd.read_csv('./datasets/plant_health.csv')
df.drop(columns=['plant_id', 'nutrient_level', 'soil_ph'],
        inplace=True, errors='ignore')  # Drop unnecessary columns
corr = df.corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm',
            square=True, cbar_kws={'shrink': .8})
plt.title('Correlation Heatmap of Plant Health Dataset')
plt.tight_layout()
plt.show()
```

```python
# 3. Boxplot
df = pd.read_csv('./datasets/plant_health.csv')
df.drop(columns=['nutrient_level', 'soil_ph'],
        inplace=True)  # Drop unnecessary columns
plt.figure(figsize=(10, 6))
sns.boxplot(hue='health', x='health', y='score', data=df,
            palette='Set1', showfliers=False, legend=False)
plt.title('Boxplot of Score by Plant Health')
plt.xlabel('Plant Health')
plt.ylabel('Score')
plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv('./datasets/plant_health.csv')
df.drop(columns=['plant_id', 'nutrient_level'],
        inplace=True)  # Drop unnecessary columns
min_temp = df['temperature'].min()
max_temp = df['temperature'].max()
min_humidity = df['humidity'].min()
max_humidity = df['humidity'].max()
min_light = df['light'].min()
max_light = df['light'].max()
min_soil_moisture = df['soil_moisture'].min()
max_soil_moisture = df['soil_moisture'].max()
print(f"Temperature: {min_temp} - {max_temp}")
print(f"Humidity: {min_humidity} - {max_humidity}")
print(f"Light: {min_light} - {max_light}")
print(f"Soil Moisture: {min_soil_moisture} - {max_soil_moisture}")
```

### Analysis of the Plant Health Dataset

#### 1. **Pairplot**
- **Description**: The pairplot visualizes pairwise relationships between numerical features in the dataset, with the `health` column used as the hue to differentiate plant health categories.
- **Observations**:
  - There are clear clusters in some feature combinations, indicating that certain features are good predictors of plant health.
  - The diagonal KDE plots show the distribution of individual features, with some features having distinct distributions for different health categories.
  - Features like `temperature`, `humidity`, and `light` seem to have strong separability between health categories.

---

#### 2. **Correlation Heatmap**
- **Description**: The heatmap shows the correlation coefficients between numerical features in the dataset.
- **Observations**:
  - Strong positive or negative correlations are highlighted in the heatmap.
  - Features like `temperature` and `humidity` may have moderate correlations with other features, indicating potential multicollinearity.
  - The `score` feature likely has a strong correlation with other features, as it is often used as a derived metric.

---

#### 3. **Boxplot**
- **Description**: The boxplot visualizes the distribution of the `score` feature across different `health` categories.
- **Observations**:
  - The median `score` varies significantly across health categories, with healthier plants generally having higher scores.
  - The spread of `score` values (interquartile range) is narrower for healthier plants, indicating more consistent scores in those categories.
  - Outliers are excluded in this plot, making it easier to observe the central tendency and variability.

---

### Summary
- The dataset shows clear relationships between features and plant health, with some features being strong indicators of health categories.
- The correlation heatmap suggests potential feature dependencies, which should be considered during model development to avoid redundancy.
- The boxplot highlights the importance of the `score` feature in distinguishing plant health, making it a key target for predictive modeling.
