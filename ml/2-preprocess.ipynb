```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import joblib


def encode_and_scale(df):
    features = df[['temperature', 'humidity',
                   'light', 'soil_moisture']]
    scaler = StandardScaler()
    scaled = scaler.fit_transform(features)

    print(f"Mean: {scaler.mean_}")
    print(f"Standard Deviation: {scaler.scale_}")
    joblib.dump(scaler, './datasets/scaler.pkl')

    # Use the raw score as the target for regression
    labels = df['score'].values

    return scaled, labels, scaler


if __name__ == '__main__':
    df = pd.read_csv('./datasets/plant_health.csv')
    X, y, scaler = encode_and_scale(df)
    pd.DataFrame(X).to_csv('./datasets/scaled_features.csv', index=False)
    pd.DataFrame(y).to_csv('./datasets/encoded_labels.csv', index=False)
```
