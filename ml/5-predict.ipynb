```python
import numpy as np
import tensorflow as tf
import joblib

interpreter = tf.lite.Interpreter(
    model_path="./models/plant_health_model.tflite")
interpreter.allocate_tensors()

input_details = interpreter.get_input_details()
output_details = interpreter.get_output_details()

# Scaled Input: [temperature, humidity, light, soil_moisture]
input_data = np.array([[-0.16102190351705578, 0.8564047482572408,
                        -0.40389551545392843, -0.15295893366284813]], dtype=np.float32)
interpreter.set_tensor(input_details[0]['index'], input_data)
interpreter.invoke()

output = interpreter.get_tensor(output_details[0]['index'])
print("Actual health score: 68.85")
print(f"Predicted health score: {output[0][0]}")

scaler = joblib.load('./datasets/scaler.pkl')

# Raw Input: [temperature, humidity, light, soil_moisture]
raw_input = np.array([[26.4901424590337, 73.99355436586002,
                     18728.72095393189, 34.87232587538428]], dtype=np.float32)

scaled_input = scaler.transform(raw_input)
scaled_input = scaled_input.reshape(1, 4)
interpreter.set_tensor(input_details[0]['index'], scaled_input)

# Invoke the interpreter
interpreter.invoke()
output = interpreter.get_tensor(output_details[0]['index'])
print("Actual health score: 68.85")
print(f"Predicted health score: {output[0][0]}")
```
